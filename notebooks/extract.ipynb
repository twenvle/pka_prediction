```python
import pandas as pd

df = pd.read_csv("../../data/known/iupac_high-confidence_v2_3.csv")
df_T25 = df[df["T"] == str(25)]
```

```python
df_T25_pka1 = df_T25[df_T25["pka_type"] == "pKaH1"]
print(len(df_T25_pka1))
```

```python
from rdkit import Chem

carboxylic_acid =Chem.MolFromSmarts("[CX3](=O)[OX2H1]")

for index, row in df_T25_pka1.iterrows():
    mol = Chem.MolFromSmiles(row["SMILES"])
    if not mol.HasSubstructMatch(carboxylic_acid):
        df_T25_pka1.drop(index, inplace=True)
df_T25_pka1_cooh = df_T25_pka1
print(df_T25_pka1_cooh)
```

```python
df_T25_pka1_cooh.to_csv("../../data/known/df_T25_pka1_cooh.csv", index=False)
```

```python
df_T25_pka1_cooh = df_T25_pka1_cooh[["SMILES", "pka_value"]]
df_T25_pka1_cooh.to_csv("../../data/known/df_T25_pka1_cooh.csv", index=False)
```

```python
df_T25_pka1_cooh["pka_value"] = pd.to_numeric(df["pka_value"], errors="coerce")
df_proceed = df_T25_pka1_cooh.groupby("SMILES", as_index=False)["pka_value"].mean()
df_proceed.to_csv("../../data/known/df_proceed.csv", index=False)
```

```python
df_proceed_v2 = df_proceed[df_proceed["pka_value"] > 1]
df_proceed_v2.to_csv("../../data/known/df_proceed_v2.csv", index=False)
```
